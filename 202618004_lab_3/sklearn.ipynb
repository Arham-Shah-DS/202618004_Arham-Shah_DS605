# Task 1 – Load and Understand the Dataset

In this task, the dataset is loaded and explored using Pandas. The target variable is `is_canceled`, and the remaining features are separated into numerical and categorical columns.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("hotel_bookings.csv")

print("First 5 Rows:")
display(df.head())

print("Shape:", df.shape)

print("\nDataset Info:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nData Types:")
display(df.dtypes)

print("\nClass Distribution:")
display(df["is_canceled"].value_counts())

plt.figure(figsize=(5,4))
sns.countplot(x="is_canceled", data=df)
plt.title("Class Distribution of is_canceled")
plt.show()

y = df["is_canceled"]
X = df.drop(columns=["is_canceled"])

num_cols = X.select_dtypes(include=["int64","float64"]).columns.tolist()
cat_cols = X.select_dtypes(include=["object"]).columns.tolist()

print("\nNumber of Numerical Columns:", len(num_cols))
print(num_cols)

print("\nNumber of Categorical Columns:", len(cat_cols))
print(cat_cols)
```

# Task 2 – Missing Values, Leakage and Outliers

This task examines missing values, removes data leakage columns, detects extreme outliers, and creates the cleaned dataset.

```python
missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum()/len(df))*100
}).sort_values("Missing Percentage", ascending=False)

display(missing)

df = df.drop(columns=["company"])
df = df.drop(columns=["reservation_status", "reservation_status_date"])

y = df["is_canceled"]
X = df.drop(columns=["is_canceled"])

num_cols = X.select_dtypes(include=["int64","float64"]).columns.tolist()
cat_cols = X.select_dtypes(include=["object"]).columns.tolist()

selected = [
    "lead_time",
    "adr",
    "stays_in_week_nights",
    "stays_in_weekend_nights"
]

plt.figure(figsize=(12,6))
df[selected].boxplot()
plt.xticks(rotation=45)
plt.show()

rows_before = len(df)
clean_df = df.copy()

for col in selected:
    Q1 = clean_df[col].quantile(0.25)
    Q3 = clean_df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    clean_df = clean_df[
        (clean_df[col] >= lower) &
        (clean_df[col] <= upper)
    ]

rows_after = len(clean_df)

print("Rows Before:", rows_before)
print("Rows After:", rows_after)
print("Rows Removed:", rows_before - rows_after)

df = clean_df
df.to_csv("hotel_bookings_cleaned.csv", index=False)
```

# Task 3 – Create Two Preprocessing Pipelines

Two preprocessing pipelines are created using KNN imputation, One-Hot Encoding, and two scaling methods: StandardScaler and MinMaxScaler.

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

X = df.drop(columns=["is_canceled"])
y = df["is_canceled"]

num_cols = X.select_dtypes(include=["int64","float64"]).columns.tolist()
cat_cols = X.select_dtypes(include=["object"]).columns.tolist()

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

num_pipeline_A = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])

preprocessor_A = ColumnTransformer([
    ("num", num_pipeline_A, num_cols),
    ("cat", cat_pipeline, cat_cols)
])

num_pipeline_B = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])

preprocessor_B = ColumnTransformer([
    ("num", num_pipeline_B, num_cols),
    ("cat", cat_pipeline, cat_cols)
])

print("Train Shape:", X_train.shape)
print("Test Shape:", X_test.shape)
```

# Task 4 – Train Two Classification Models

Logistic Regression and Decision Tree classifiers are trained using both preprocessing pipelines.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

lr_A = Pipeline([
    ("preprocessor", preprocessor_A),
    ("model", LogisticRegression(max_iter=1000))
])

lr_B = Pipeline([
    ("preprocessor", preprocessor_B),
    ("model", LogisticRegression(max_iter=1000))
])

dt_A = Pipeline([
    ("preprocessor", preprocessor_A),
    ("model", DecisionTreeClassifier(random_state=42))
])

dt_B = Pipeline([
    ("preprocessor", preprocessor_B),
    ("model", DecisionTreeClassifier(random_state=42))
])

lr_A.fit(X_train, y_train)
lr_B.fit(X_train, y_train)
dt_A.fit(X_train, y_train)
dt_B.fit(X_train, y_train)

print("All four models trained successfully.")
```

# Task 5 – Evaluate and Compare the Four Results

The four model-pipeline combinations are evaluated using accuracy, precision, recall, F1-score, and confusion matrices.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import pandas as pd
import matplotlib.pyplot as plt

def evaluate(model, name):
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    return {
        "Model": name,
        "Train Accuracy": accuracy_score(y_train, train_pred),
        "Test Accuracy": accuracy_score(y_test, test_pred),
        "Precision": precision_score(y_test, test_pred),
        "Recall": recall_score(y_test, test_pred),
        "F1 Score": f1_score(y_test, test_pred)
    }

results = pd.DataFrame([
    evaluate(lr_A, "Logistic + StandardScaler"),
    evaluate(lr_B, "Logistic + MinMaxScaler"),
    evaluate(dt_A, "Decision Tree + StandardScaler"),
    evaluate(dt_B, "Decision Tree + MinMaxScaler")
])

results["Train-Test Gap"] = results["Train Accuracy"] - results["Test Accuracy"]

display(results.round(4))

best_lr = lr_A if evaluate(lr_A,"")["Test Accuracy"] >= evaluate(lr_B,"")["Test Accuracy"] else lr_B

ConfusionMatrixDisplay(confusion_matrix(y_test, best_lr.predict(X_test))).plot(cmap="Blues")
plt.title("Best Logistic Regression")
plt.show()

best_dt = dt_A if evaluate(dt_A,"")["Test Accuracy"] >= evaluate(dt_B,"")["Test Accuracy"] else dt_B

ConfusionMatrixDisplay(confusion_matrix(y_test, best_dt.predict(X_test))).plot(cmap="Greens")
plt.title("Best Decision Tree")
plt.show()
```
