<h2 align="center">Part A - Vectorized Programming with NumPy<br>
Task 1 - Arrays, Statistics, and Indexing
</h2>

```python
import numpy as np

rng = np.random.default_rng(42)
array_A = rng.integers(low=10, high=100, size=100)
print("Array A:\n", array_A)

print("Min:", array_A.min())
print("Max:", array_A.max())
print("Median:", np.median(array_A))
print("Mean:", array_A.mean())
print("Std dev:", array_A.std())

array_B = np.arange(0, 100)
print("\nArray B (First 10 elements):", array_B[:10])

zeros_arr = np.zeros(100)
ones_arr = np.ones(100)
print("\nZeros Shape:", zeros_arr.shape, "| Data Type:", zeros_arr.dtype)
print("Ones Shape:", ones_arr.shape, "| Data Type:", ones_arr.dtype)

linspace_arr = np.linspace(0, 10, 100)
print("\nLinspace Array (First 5 elements):", linspace_arr[:5])

arr_2d = np.arange(1, 13).reshape(3, 4)
print("\n2D Array:\n", arr_2d)
print("2D Shape:", arr_2d.shape, "| ndim:", arr_2d.ndim)
print("Row 0:", arr_2d[0, :])
print("Col 1:", arr_2d[:, 1])
print("Specific element:", arr_2d)
print("Slicing (Top-left 2x2):\n", arr_2d[0:2, 0:2])

arr_3d = np.arange(1, 25).reshape(2, 3, 4)
print("\n3D Array:\n", arr_3d)
print("3D Shape:", arr_3d.shape, "| ndim:", arr_3d.ndim)
print("Specific element:", arr_3d)
print("Slicing (First row of all matrices):\n", arr_3d[:, 0, :])

matrix_orig = np.arange(10, 16).reshape(2, 3)
print("\nReshaped 2x3 Matrix:\n", matrix_orig)
matrix_flat = matrix_orig.flatten()
print("Flattened back to 1D:", matrix_flat)
```

<h2>Task 2 - Vectorized Arithmetic and Linear Algebra
</h2>

```python
import numpy as np

M1 = np.array([[1, 2], [3, 4]])
M2 = np.array([[5, 6], [7, 8]])

print("Matrix Addition:\n", M1 + M2)
print("Element-wise Multiplication:\n", M1 * M2)
print("Matrix Multiplication (@):\n", M1 @ M2)

sq_matrix = np.array([[4, 7], [2, 6]])
print("\nSquare Matrix:\n", sq_matrix)
print("Transpose:\n", sq_matrix.T)
det = np.linalg.det(sq_matrix)
print("Determinant:", det)

if det != 0:
    sq_inv = np.linalg.inv(sq_matrix)
    print("Inverse Matrix:\n", sq_inv)
    is_verified = np.allclose(sq_matrix @ sq_inv, np.eye(2))
    print("Verification (A @ A^-1 == I):", is_verified)
```

<h2>Task 3 - Normal Distribution and Histogram</h2>


```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
true_mean = 50
true_std = 15
norm_vals = rng.normal(loc=true_mean, scale=true_std, size=1000)

sample_mean = norm_vals.mean()
sample_std = norm_vals.std()

print(f"Theoretical Mean: {true_mean} | Sample Mean: {sample_mean:.4f}")
print(f"Theoretical Std Dev: {true_std} | Sample Std Dev: {sample_std:.4f}")

plt.figure(figsize=(7, 4.5))
plt.hist(norm_vals, bins=30, edgecolor='black', color='skyblue', alpha=0.7)
plt.axvline(sample_mean, color='red', linestyle='dashed', linewidth=1.5, label=f'Sample Mean ({sample_mean:.2f})')
plt.title('Task 3: Normal Distribution')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.show()
```

<h1>Part B - Data Wrangling with Pandas: Titanic Dataset</h1>
<h2>Task 4 - Load and Inspect Data</h2>

```python
import pandas as pd

df = pd.read_csv("titanic/train.csv")

print("First 5 rows:")
display(df.head())

print("Last 5 rows:")
display(df.tail())

print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nUsing loc:")
display(df.loc[0:4, ['Name', 'Age', 'Survived']])

print("\nUsing iloc:")
display(df.iloc[0:5, [0, 1, 2]])

print("\nDifference:")
print("loc selects data using labels/names.")
print("iloc selects data using integer positions.")
```

<h2>Task 5 - Filtering and Querying</h2>

```python
male_above_50 = df[(df['Sex'] == 'male') & (df['Age'] > 50)]
print("1. Male passengers older than 50:", len(male_above_50))

female_first_class = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
survival_percentage = female_first_class['Survived'].mean() * 100
print("2. Female first-class passengers:", len(female_first_class))
print("   Percentage survived:", round(survival_percentage, 2), "%")

median_fare = df['Fare'].median()
age_20_40 = df[(df['Age'] >= 20) & (df['Age'] <= 40) & (df['Fare'] > median_fare) & (df['Survived'] == 1)]
print("3. Age 20-40, fare above median, and survived:", len(age_20_40))

alone_under_30 = df[(df['SibSp'] == 0) & (df['Parch'] == 0) & 
                    (df['Age'] < 30) & (df['Survived'] == 0)]
print("4. Travelling alone, age < 30, and did not survive:", len(alone_under_30))

southampton_median = df[df['Embarked'] == 'S']['Fare'].median()
southampton_filter = df[(df['Embarked'] == 'S') & 
                        (df['Pclass'].isin([2, 3])) & 
                        (df['Fare'] > southampton_median)]
print("5. Embarked S, Pclass 2 or 3, fare above Southampton median:", len(southampton_filter))
```

<h2>Task 6 - groupby() and Aggregation
</h2>

```python
print("1. Survival rate by Sex:")
display(df.groupby('Sex')['Survived'].mean().mul(100).round(2))

print("\n2. Survival rate by Pclass:")
display(df.groupby('Pclass')['Survived'].mean().mul(100).round(2))

print("\n3. Average Age and Fare by Pclass:")
display(df.groupby('Pclass')[['Age', 'Fare']].mean().round(2))

print("\n4. Passenger count and survival rate by Sex-Pclass:")
display(
    df.groupby(['Sex', 'Pclass'])
      .agg(
          Passenger_Count=('PassengerId', 'count'),
          Survival_Rate=('Survived', 'mean')
      )
      .assign(Survival_Rate=lambda x: (x['Survival_Rate'] * 100).round(2))
)

print("\n5. Passenger count, average Fare, and survival rate by Embarked:")
display(
    df.groupby('Embarked')
      .agg(
          Passenger_Count=('PassengerId', 'count'),
          Average_Fare=('Fare', 'mean'),
          Survival_Rate=('Survived', 'mean')
      )
      .assign(
          Average_Fare=lambda x: x['Average_Fare'].round(2),
          Survival_Rate=lambda x: (x['Survival_Rate'] * 100).round(2)
      )
)
```

<h2>Task 7 - Missing Values and Fare Outliers</h2>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("1. Missing-value count and percentage:")

missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df) * 100).round(2)

missing_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage': missing_percentage
})

display(missing_summary)

plt.figure(figsize=(8, 5))
df.isnull().sum().plot(kind='bar', edgecolor='black')
plt.title('Missing Values by Column')
plt.xlabel('Columns')
plt.ylabel('Missing Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


print("2. Fill missing Age values with mean:")

print("Missing Age before:", df['Age'].isnull().sum())

age_mean = df['Age'].mean()
df['Age'] = df['Age'].fillna(age_mean)

print("Missing Age after:", df['Age'].isnull().sum())


print("\n3. Imputation methods:")

df_impute = pd.read_csv("titanic/train.csv")

mean_age = df_impute['Age'].mean()
median_age = df_impute['Age'].median()
mode_age = df_impute['Age'].mode()[0]

df_impute['Age_Mean'] = df_impute['Age'].fillna(mean_age)
df_impute['Age_Median'] = df_impute['Age'].fillna(median_age)
df_impute['Age_Mode'] = df_impute['Age'].fillna(mode_age)

random_value = np.random.choice(df_impute['Age'].dropna())
df_impute['Age_Random'] = df_impute['Age'].fillna(random_value)

print("Mean:", mean_age)
print("Median:", median_age)
print("Mode:", mode_age)
print("Random value used:", random_value)

print("\nMissing values after imputation:")
print("Mean:", df_impute['Age_Mean'].isnull().sum())
print("Median:", df_impute['Age_Median'].isnull().sum())
print("Mode:", df_impute['Age_Mode'].isnull().sum())
print("Random:", df_impute['Age_Random'].isnull().sum())


print("\n4. Fare Outliers:")

Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", len(outliers))
```

<h2>Task 8 - Features and Pivot Table</h2>

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

print("FamilySize and IsAlone:")
display(df[['SibSp', 'Parch', 'FamilySize', 'IsAlone']].head())

pivot = pd.pivot_table(
    df,
    values='Survived',
    index='Sex',
    columns='Pclass',
    aggfunc='mean'
)

print("\nPivot Table:")
display(pivot)

highest_group = pivot.stack().idxmax()
highest_value = pivot.stack().max()

lowest_group = pivot.stack().idxmin()
lowest_value = pivot.stack().min()

print("Highest survival group:", highest_group)
print("Highest survival rate:", round(highest_value * 100, 2), "%")

print("Lowest survival group:", lowest_group)
print("Lowest survival rate:", round(lowest_value * 100, 2), "%")
```

<h2>Task 9 - Visualizations and Observations</h2>

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

numerical_columns = [
    'Survived', 'Pclass', 'Age', 'SibSp',
    'Parch', 'Fare', 'FamilySize', 'IsAlone'
]

correlation = df[numerical_columns].corr()

plt.figure(figsize=(9, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()

print("Strongest positive relationship:")
print(correlation.where(correlation != 1).stack().idxmax(),
      round(correlation.where(correlation != 1).stack().max(), 2))

print("\nStrongest negative relationship:")
print(correlation.where(correlation != 1).stack().idxmin(),
      round(correlation.where(correlation != 1).stack().min(), 2))


survival_by_sex = df.groupby('Sex')['Survived'].mean() * 100

plt.figure(figsize=(6, 4))
survival_by_sex.plot(kind='bar', edgecolor='black')
plt.title('Survival Rate by Sex')
plt.xlabel('Sex')
plt.ylabel('Survival Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("Male survival rate:", round(survival_by_sex['male'], 2), "%")
print("Female survival rate:", round(survival_by_sex['female'], 2), "%")


plt.figure(figsize=(8, 5))

survived = df[df['Survived'] == 1]
not_survived = df[df['Survived'] == 0]

plt.scatter(not_survived['Age'], not_survived['Fare'], label='Survived = 0', alpha=0.5)
plt.scatter(survived['Age'], survived['Fare'], label='Survived = 1', alpha=0.5)

plt.title('Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend()
plt.tight_layout()
plt.show()
```
